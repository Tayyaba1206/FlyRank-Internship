## 1. Ranked actions + reason codes

The action queue prioritizes items where intervention may have the highest value.

Each recommendation includes a reason code so a human can understand why the item was ranked.

Example reason codes:

- HIGH_RISK: The item shows signals that require early attention.
- DELAY_PATTERN: Historical patterns indicate possible delays.
- LOW_CONFIDENCE: The prediction is uncertain and requires review.
- NORMAL: No urgent action recommended.

The ranking is decision-support only. It does not replace human judgement.

```python
# Example action queue

import pandas as pd

action_queue = pd.DataFrame({
    "Rank": [1, 2, 3, 4],
    "Item": ["Order_A", "Order_B", "Order_C", "Order_D"],
    "Priority": ["High", "High", "Medium", "Low"],
    "Reason_Code": [
        "HIGH_RISK",
        "DELAY_PATTERN",
        "LOW_CONFIDENCE",
        "NORMAL"
    ]
})

action_queue
```

## 2. Intended use and limits

### Intended Use

This system is designed to support teams in prioritizing content actions and identifying items that may need attention.

Possible users:

- Decision makers monitoring trends

### Limitations

The system should not be used as the only source for important decisions.

It may become inaccurate when:

- Data patterns change
- New situations appear
- Input data quality decreases
- Human context is missing

The output provides recommendations, not automatic decisions.

## 3. Human review + the no-go list

## 3. Human review + the no-go list

Before taking action, a human reviewer should check:

- Whether the recommendation matches the real situation
- Whether input data is complete and correct
- Whether there are external factors not captured by the system
- Whether the action could create unexpected impact


## No-go list

The system should never:

- Automatically make final decisions without human approval
- Replace expert judgement
- Use private or sensitive information without permission
- Create actions that affect people without review
- Treat predictions as guaranteed outcomes
