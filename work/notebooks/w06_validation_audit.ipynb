## Leakage Audit

I reviewed all features used in the final model to identify possible data leakage.

The following columns were excluded because they may contain information related to the prediction target or future outcomes:

- trend_pct
- impression_tier
- position_tier
- trend_direction (used only as the target)

The remaining features (search_volume, competition, cpc, word_count, and char_count) are available before the prediction is made. This keeps the model suitable for decision-support rather than using future information.

```python
leakage_columns = [
    "trend_pct",
    "impression_tier",
    "position_tier"
]

print("Potential leakage columns:")
for col in leakage_columns:
    print("-", col)

print("\nModel features:")
print(features)

safe = all(col not in features for col in leakage_columns)

print("\nLeakage Found:", not safe)
```
