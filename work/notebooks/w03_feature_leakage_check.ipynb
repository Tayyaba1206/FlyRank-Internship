## Leakage Check

The field **trend_pct** was intentionally tested.

It directly determines the target (trend_direction), so using it would leak future information and produce unrealistic model performance.

Therefore, it was removed from the final feature vector.

```python
print("Potential Leakage Column:")
print(df["trend_pct"].head())

print("\nTarget:")
print(df["trend_direction"].head())
```
