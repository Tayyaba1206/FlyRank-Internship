## Method Choice

For this project, I selected the Random Forest Classifier. It is suitable because it can learn complex relationships between multiple observable features and is more flexible than a simple rule-based approach. I will compare its performance against the Week 4 baseline using the same dataset and evaluation method.

```python
import pandas as pd

from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score

import os
import sys
import subprocess

# Setup
if not os.path.exists("data/raw/content_refresh_anonymized.csv"):
    REPO = "flyrank-ml-internship-starter"

    if not os.path.exists(REPO):
        subprocess.run([
            "git","clone",
            "https://github.com/flyrank-bih/flyrank-ml-internship-starter.git"
        ])

    os.chdir(REPO)

df = pd.read_csv("data/raw/content_refresh_anonymized.csv")

print(df.shape)
```

## 2. Split design

*Grouped by client? Time-aware? Say why this split is honest for your question.*

## Split Design

I used an 80/20 train-test split. The training data is used to train the model, while the testing data is reserved for evaluation. This provides an honest comparison between the baseline rule and the machine learning model.

```python
features = [
    "search_volume",
    "word_count",
    "impressions_90d",
    "ctr",
    "avg_position",
    "days_since_last_update"
]

df["target"] = (
    df["trend_direction"]=="down"
).astype(int)

X = df[features].fillna(0)
y = df["target"]

X_train,X_test,y_train,y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42
)

print(X_train.shape)
print(X_test.shape)
```

## 3. Train + compare vs my baseline

*Same data, same metric, same split as your Week-4 baseline. Show the table.*

## Model vs Baseline

The Random Forest model is compared with the Week 4 baseline using the same dataset split. The comparison is based on prediction accuracy and classification performance.

```python
model = RandomForestClassifier(
    random_state=42
)

model.fit(
    X_train,
    y_train
)

pred = model.predict(
    X_test
)

accuracy = accuracy_score(
    y_test,
    pred
)

print("Random Forest Accuracy:",accuracy)

baseline_accuracy = 0.68

comparison = pd.DataFrame({
    "Method":[
        "Baseline Rule",
        "Random Forest"
    ],
    "Accuracy":[
        baseline_accuracy,
        accuracy
    ]
})

comparison
```
